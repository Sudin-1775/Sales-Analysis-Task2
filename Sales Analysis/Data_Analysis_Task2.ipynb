# Task 2: Sales Data Analysis

This notebook loads the cleaned sales dataset and performs exploratory and aggregated sales data analysis, including KPI calculations, regional and product performance, and trend visualizations.

### 1. Load Cleaned Dataset
Load the cleaned Excel file generated in Task 1.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

cleaned_file_path = "../01_Data_Cleaning/CodeOrbit_Sales_Cleaned.xlsx"
if not os.path.exists(cleaned_file_path):
    cleaned_file_path = "CodeOrbit_Sales_Cleaned.xlsx"

df = pd.read_excel(cleaned_file_path)
print("Cleaned dataset loaded successfully!")
print(f"Dataset Shape: {df.shape}")
display(df.head())
```

### 2. Calculate Key Performance Indicators (KPIs)

```python
total_sales = df["Sales"].sum()
total_quantity = df["Quantity"].sum()
average_order_value = df["Sales"].mean()

print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Average Order Value: ₹{average_order_value:,.2f}")
```

### 3. Sales Grouped Aggregations
Analyze top products, sales by category, and regional performance.

```python
print("
--- Top 10 Products by Sales ---")
top_products = df.groupby("Product")["Sales"].sum().sort_values(ascending=False).head(10)
print(top_products)

print("
--- Sales by Category ---")
sales_by_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(sales_by_category)

print("
--- Sales by Region ---")
sales_by_region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(sales_by_region)
```

### 4. Sales Trends over Time

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Month"] = df["Order_Date"].dt.to_period("M")
monthly_sales = df.groupby("Month")["Sales"].sum()

print("--- Monthly Sales Trend ---")
print(monthly_sales)
```

### 5. Sales Visualizations

```python

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(ax=axes[0], x=sales_by_category.index, y=sales_by_category.values, hue=sales_by_category.index, legend=False, palette="viridis")
axes[0].set_title("Sales by Category")
axes[0].set_ylabel("Total Sales ($)")
axes[0].tick_params(axis='x', rotation=45)

sns.barplot(ax=axes[1], x=sales_by_region.index, y=sales_by_region.values, hue=sales_by_region.index, legend=False, palette="plasma")
axes[1].set_title("Sales by Region")
axes[1].set_ylabel("Total Sales ($)")
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
plt.figure(figsize=(12, 5))
monthly_sales.index = monthly_sales.index.astype(str)
sns.lineplot(x=monthly_sales.index, y=monthly_sales.values, marker="o", linewidth=2.5)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
